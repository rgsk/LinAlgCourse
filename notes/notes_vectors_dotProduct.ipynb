```python
import numpy as np
```

```python
v1 = np.array([1, 2, 3])
v2 = np.array([3, 2, 5])

np.multiply(v1, v2)
```

```python
dp1 = sum(np.multiply(v1, v2))
dp1
```

```python
dp2 = np.dot(v1, v2)
dp2
```

```python
dp3 = np.matmul(v1, v2)
dp3
```

```python
dp4 = 0

for i in range(len(v1)):
    dp4 = dp4 + v1[i] * v2[i]
dp4
```
