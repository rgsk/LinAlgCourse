# Geometric Vectors — Summary Notes

## What Is a Geometric Vector?

A **geometric vector** represents a **displacement** (movement), defined by:

- **Direction**
- **Length (magnitude)**

A vector does **not** represent a fixed position — it represents **how to move**.

Example:

Vector **[1, 2]** means:

- Move **1 unit right** (x-direction)
- Move **2 units up** (y-direction)

---

## Same Vector in Different Locations

Many arrows can represent the **same vector**.

All arrows that:

- Move the **same horizontal amount**
- Move the **same vertical amount**
- Have the **same direction**
- Have the **same length**

represent the **same vector**, even if they start at different points.

Example:

These are all the same vector **[1,2]**:

- From (0,0) → (1,2)  
- From (2,1) → (3,3)  
- From (-1,-2) → (0,0)

They all move:

- **+1 in x**
- **+2 in y**

So they represent the **same displacement**.

---

## Vector in Standard Position

A vector is in **standard position** when:

- Its **tail starts at the origin (0,0)**
- Its **head ends at the point given by its components**

Example:

Vector **[1,2]** in standard position:

- Starts at **(0,0)**
- Ends at **(1,2)**

So:

**Vector [1,2] in standard position corresponds to the point (1,2).**

---

## Vector vs Point (Important Distinction)

This is a key conceptual difference.

| Object | Meaning |
|-------|--------|
| **Point (1,2)** | A **location** in space |
| **Vector [1,2]** | A **movement**: 1 right, 2 up |

However:

- A vector in **standard position** ends at a point
- That’s why vectors and coordinates can look similar

But conceptually:

**Points describe position**  
**Vectors describe displacement**

---

## Key Takeaways

- A **vector** is defined by **direction and magnitude**, not by where it starts.
- The same vector can be **shifted anywhere** without changing its value.
- A vector in **standard position** starts at **(0,0)**.
- Vector **[a,b]** in standard position ends at **(a,b)**.
- Vectors represent **movement**, while points represent **locations**.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 2-dimensional vector
v2 = [3, -2]

# Create a fresh 2D axes so this cell is not affected by any previous 3D plot.
fig, ax = plt.subplots(figsize=(5, 5))

# Plot the 2D vector from (0, 0) to (v2[0], v2[1]).
ax.plot(
    [0, v2[0]],
    [0, v2[1]],
)

# Draw the x-axis and y-axis.
ax.plot(
    [-4, 4],
    [0, 0],
    "k--",
)
ax.plot(
    [0, 0],
    [-4, 4],
    "k--",
)

ax.grid()
ax.axis((-4, 4, -4, 4))  # (xmin, xmax, ymin, ymax)
plt.show()
```

```python
v3 = [4, -3, 2]
# plot the 3D vector
fig = plt.figure(figsize=plt.figaspect(1))
ax = fig.add_subplot(projection="3d")
ax.plot([0, v3[0]], [0, v3[1]], [0, v3[2]], linewidth=2)

# make the plot look nicer
ax.plot([0, 0], [0, 0], [-4, 4], "k--")
ax.plot([0, 0], [-4, 4], [0, 0], "k--")
ax.plot([-4, 4], [0, 0], [0, 0], "k--")
plt.show()
```

```python
# Interactive 3D vector plot: run this cell, then drag the plot to rotate it
%matplotlib widget

v3_interactive = np.array(v3)
limit = 5

fig = plt.figure(figsize=(6, 6))
ax = fig.add_subplot(111, projection="3d")

# Draw the vector as an arrow from the origin.
ax.quiver(
    0, 0, 0, *v3_interactive, color="tab:blue", linewidth=3, arrow_length_ratio=0.12
)

# Add coordinate axes for orientation.
ax.plot([-limit, limit], [0, 0], [0, 0], "k--", alpha=0.45)
ax.plot([0, 0], [-limit, limit], [0, 0], "k--", alpha=0.45)
ax.plot([0, 0], [0, 0], [-limit, limit], "k--", alpha=0.45)

ax.set_xlim(-limit, limit)
ax.set_ylim(-limit, limit)
ax.set_zlim(-limit, limit)
ax.set_box_aspect((1, 1, 1))
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")
ax.set_title(f"3D vector v3 = {v3_interactive.tolist()}")
ax.view_init(elev=20, azim=35)

plt.show()
```

```python
# Interactive 3D vector plot with Blender-style view buttons
# Run this cell, then drag the plot to rotate it manually.
%matplotlib widget

import ipywidgets as widgets
from IPython.display import display

v3_interactive = np.array(v3)
limit = 5

fig = plt.figure(figsize=(6, 6))
ax = fig.add_subplot(111, projection="3d")

# Draw the vector as an arrow from the origin.
ax.quiver(
    0,
    0,
    0,
    *v3_interactive,
    color="tab:blue",
    linewidth=3,
    arrow_length_ratio=0.12,
)

# Add coordinate axes for orientation.
ax.plot([-limit, limit], [0, 0], [0, 0], "r--", alpha=0.55, label="x-axis")
ax.plot([0, 0], [-limit, limit], [0, 0], "g--", alpha=0.55, label="y-axis")
ax.plot([0, 0], [0, 0], [-limit, limit], "b--", alpha=0.55, label="z-axis")

ax.set_xlim(-limit, limit)
ax.set_ylim(-limit, limit)
ax.set_zlim(-limit, limit)
ax.set_box_aspect((1, 1, 1))
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")
ax.set_title(f"3D vector v3 = {v3_interactive.tolist()}")
ax.legend(loc="upper left")

# These are camera positions, similar to choosing front/top/side views in Blender.
views = {
    "3D": (20, 35),  # angled view
    "XY": (90, -90),  # x-y plane, looking along the z-axis
    "XZ": (0, -90),  # x-z plane, looking along the y-axis
    "YZ": (0, 0),  # y-z plane, looking along the x-axis
    "+X": (0, 0),  # look from the positive x direction
    "-X": (0, 180),  # look from the negative x direction
    "+Y": (0, 90),  # look from the positive y direction
    "-Y": (0, -90),  # look from the negative y direction
    "+Z": (90, -90),  # look from above
    "-Z": (-90, -90),  # look from below
}

view_buttons = []

for label, (elev, azim) in views.items():
    button = widgets.Button(description=label, layout=widgets.Layout(width="55px"))

    def set_view(_, elev=elev, azim=azim, label=label):
        ax.view_init(elev=elev, azim=azim)
        ax.set_title(f"{label} view: v3 = {v3_interactive.tolist()}")
        fig.canvas.draw_idle()

    button.on_click(set_view)
    view_buttons.append(button)

ax.view_init(*views["3D"])
display(widgets.HBox(view_buttons))
plt.show()
```
