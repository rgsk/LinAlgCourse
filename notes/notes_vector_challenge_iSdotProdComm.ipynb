```python
import numpy as np
```

```python
a = np.random.randn(100)
b = np.random.randn(100)
```

```python
np.dot(a, b)
```

```python
np.dot(b, a)
```

```python
np.dot(a, b) == np.dot(b, a)
```

```python
a = np.array([1, 2])
b = np.array([3, 4])
```

```python
np.dot(a, b)
```

```python
np.dot(b, a)
```
