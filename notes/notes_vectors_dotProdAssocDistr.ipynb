```python
import numpy as np
```

```python
## Distributive property

# create random vectors
n = 10
a = np.random.randn(n)
b = np.random.randn(n)
c = np.random.randn(n)

# the two results
res1 = np.dot(a, (b + c))
res2 = np.dot(a, b) + np.dot(a, c)

# compare them
print([res1, res2])
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
c = np.array([7, 8, 9])
print([np.dot(a, b + c), np.dot(a, b) + np.dot(a, c)])
```

```python
## Associative property

# create random vectors
n = 5
a = np.random.randn(n)
b = np.random.randn(n)
c = np.random.randn(n)


def test_associative(a, b, c):
    # the two results
    res1 = np.dot(a, np.dot(b, c))
    res2 = np.dot(np.dot(a, b), c)

    # compare them
    print(res1)
    print(res2)


test_associative(a, b, c)
```

```python
### special cases where associative property works!
# 1) one vector is the zeros vector
# 2) a==b==c

a = np.zeros(n)
test_associative(a, b, c)
```

```python
a = b = c = np.array([2, 3, 4])
test_associative(a, b, c)
```
